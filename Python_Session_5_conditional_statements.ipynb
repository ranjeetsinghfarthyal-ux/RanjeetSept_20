```python
# Im entering to bus
# conductor is asking how much distnce is travel
# if the distance more than 25km then i need to pay charge
# otherwise free ride

distance=eval(input('enter a distance:')) #100
if distance>=25:  # 100>=25
    charge=eval(input('enter the charge:')) # 2
    charg_dist=distance-25 # 100-25=75
    total_charge=charg_dist*charge
    print(f'The total charge is:{total_charge}')
else:
    print('enjoy the free ride')
```

```python
# ask the user enter  a percentage
# if percentage greater than or equal to 75 then print A
# if perentage between 60 to 74 then print B
# if percentage between 45 to 59 then print C
# if perecentage between 35 to 44 then print D
# otherwise print fail
```

```python
percent=eval(input('enter a percentage')) # 90
if percent>=75:  # 65>=75  F
   print('print A')
elif percent >=60:  # 90>=60  T 
   print('print B')
elif percent >=45:   # 90>=45  T 
   print('print C')
elif percent >=35:   # 90>=35 T
   print('print D')
else:
   print('fail')
```
